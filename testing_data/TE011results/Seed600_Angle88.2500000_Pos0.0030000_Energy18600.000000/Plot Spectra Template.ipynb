```python
import ROOT
```

```python
c = ROOT.TCanvas("c")
```

```python
fPS = ROOT.TFile("basic_Angle88.2500000_Pos0.0030000_Energy18600.000000.root")
```

```python
fPS.ls()
```

```python
hPS = fPS.Get("histPSD_0_0")
hPS.Draw()
c.Draw()
```

```python
# Assume fPS and hPS are already defined
#c = ROOT.TCanvas("c", "Canvas", 800, 600)
hPS.Draw()

# Peak search
spectrum = ROOT.TSpectrum()
nPeaks = spectrum.Search(hPS, 2, "", 0.05)  # (hist, sigma, options, threshold)

print(f"Found {nPeaks} peaks:")
peaks = spectrum.GetPositionX()  # array of peak x-positions

for i in range(nPeaks):
    print(f"Peak {i+1}: x = {peaks[i]}")
```

```python
# Maximum y value (bin content)
y_max = hPS.GetMaximum()
print("Maximum y value =", y_max)

# Location of that maximum (x value)
max_bin = hPS.GetMaximumBin()
x_at_max = hPS.GetBinCenter(max_bin)

print(f"Peak at x = {x_at_max}, y = {y_max}")
```
